```python
import pandas as pd

import matplotlib.pyplot as plt

# Create ADR ranges

df['ADR_Range'] = pd.cut(

    df['adr'],

    bins=[0, 50, 100, 150, 200, 300, 500, 1000],

    labels=['0-50', '50-100', '100-150', '150-200',

            '200-300', '300-500', '500-1000']

)

# Count bookings in each range

adr_counts = df['ADR_Range'].value_counts().sort_index()

# Bar chart

adr_counts.plot(kind='bar')

plt.xlabel('ADR Range')

plt.ylabel('Number of Bookings')

plt.title('Distribution of ADR')

plt.xticks(rotation=45)

plt.show()
```

```python
Q1 = df['adr'].quantile(0.25)
Q3 = df['adr'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers = df[
    (df['adr'] < lower_limit) |
    (df['adr'] > upper_limit)
]

print("Number of outliers:", len(outliers))
print("Outlier ADR values:")
print(outliers['adr'].sort_values().to_list())
```

```python
import matplotlib.pyplot as plt
```

```python
df = df[
    (df['adr'] >= lower_limit) &
    (df['adr'] <= upper_limit)
]
print("Dataset after removing outliers:", df.shape)
```

```python
# Box plot
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.boxplot(df['adr'].dropna())
plt.ylabel('ADR')
plt.title('ADR After Removing Outliers')
plt.show()
```
